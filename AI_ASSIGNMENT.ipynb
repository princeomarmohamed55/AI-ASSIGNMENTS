3. Data Collection & Preparation
Data cleaning steps are applied prior to model training to prevent data leakage:

Dropped the non-predictive `id` column and filtered out the single 'Other' gender record.
Split data using stratified sampling before imputation and scaling.
Constructed a `ColumnTransformer` pipeline handling median imputation for numerical features and one-hot encoding for categorical variables.

```python
# Clean data and drop 'Other' gender
clean_df = df[df['gender'] != 'Other'].drop(columns='id')

# Features and target split
X = clean_df.drop(columns='stroke')
y = clean_df['stroke']

# Stratified Train-Test Split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.1, stratify=y, random_state=42
)

# Define preprocessing column transformer
num_cols = ['age', 'hypertension', 'heart_disease', 'avg_glucose_level', 'bmi']
cat_cols = ['gender', 'ever_married', 'work_type', 'Residence_type', 'smoking_status']

preprocess = ColumnTransformer([
    ('num', Pipeline([('imputer', SimpleImputer(strategy='median')), ('scaler', StandardScaler())]), num_cols),
    ('cat', OneHotEncoder(drop='first', handle_unknown='ignore'), cat_cols)
])

# Visual: BMI before vs after median imputation
bmi_imp = X_train['bmi'].fillna(X_train['bmi'].median())

fig, ax = plt.subplots(1, 2, figsize=(13, 5))

sns.histplot(X_train['bmi'].dropna(), bins=25, color='teal', ax=ax[0])
ax[0].set_title('BMI before imputation')

sns.histplot(bmi_imp, bins=25, color='orange', ax=ax[1])
ax[1].set_title('BMI after median imputation')

plt.tight_layout()
plt.savefig('figures/04_bmi_before_after.png', dpi=150)
plt.show()
```

4–5. Model Selection & Training (Logistic Regression)
Iterative refinement from baseline parameters to tuned optimization:

```python
def evaluate(model, name, threshold=0.45):
    proba = model.predict_proba(X_test)[:, 1]
    pred = (proba >= threshold).astype(int)

    res = dict(model=name, accuracy=accuracy_score(y_test, pred),
               precision=precision_score(y_test, pred, zero_division=0),
               recall=recall_score(y_test, pred),
               f1=f1_score(y_test, pred),
               roc_auc=roc_auc_score(y_test, proba))

    print(classification_report(y_test, pred,
          target_names=['No stroke', 'Stroke'], zero_division=0))

    return res, pred, proba


baseline = Pipeline([
    ('prep', preprocess),
    ('clf', LogisticRegression(max_iter=1500))
]).fit(X_train, y_train)

res_base, pred_base, proba_base = evaluate(baseline, 'Baseline LR')

ConfusionMatrixDisplay.from_predictions(
    y_test, pred_base, cmap='Purples', values_format='d'
)

plt.title('Baseline Model Confusion Matrix')
plt.savefig('figures/05_cm_baseline.png', dpi=180)
plt.show()
```

refinement 1

```python
# Define and train the class-weighted logistic regression model
balanced = Pipeline([
    ('prep', preprocess),
    ('clf', LogisticRegression(max_iter=1500, class_weight='balanced'))
]).fit(X_train, y_train)

res_bal, pred_bal, proba_bal = evaluate(balanced, 'LR + class_weight')

ConfusionMatrixDisplay.from_predictions(y_test, pred_bal, cmap='Greens')
plt.title('Class-weighted confusion matrix')

plt.savefig('figures/06_cm_balanced.png', dpi=180)
plt.show()
```

Refinement 2: Hyperparameter tuning using 5-fold Stratified Cross-Validation (`GridSearchCV`) across regularization strengths (`C`) and penalty types (`l1`, `l2`).

6. Test & Present Results
Performance metrics across model iterations demonstrate progressive gains in recall for the minority stroke class:

```python
pipe = Pipeline([('prep', preprocess), ('clf',
LogisticRegression(max_iter=1500, class_weight='balanced'))])

grid = {'clf__C':[0.05, 0.5, 2, 20, 50], 'clf__penalty':['l1','l2'],
'clf__solver':['liblinear']}

gs = GridSearchCV(pipe, grid, scoring='roc_auc', cv=StratifiedKFold(4,
shuffle=True, random_state=21), n_jobs=-1)

gs.fit(X_train, y_train)

print('Best params:', gs.best_params_, '| CV ROC-AUC:',
round(gs.best_score_,3))

best = gs.best_estimator_

res_tuned, pred_tuned, proba_tuned = evaluate(best, 'Tuned LR')
ConfusionMatrixDisplay.from_predictions(y_test, pred_tuned, cmap='Oranges')
plt.title('Tuned Logistic Regression Confusion Matrix')
plt.tight_layout()
plt.savefig('figures/07_cm_tuned.png', dpi=180)
plt.show()
```

```python
results = pd.DataFrame([res_base, res_bal, res_tuned]).set_index('model').round(2)

results.to_csv('results_summary.csv')
results
```

7. Evaluation & Recommended Improvements
Discussion Points:

Tradeoffs: In a clinical screening context, prioritizing high recall minimizes false negatives at the cost of lowered precision (more false alarms requiring follow-up).
Model Limitations: Scarcity of positive samples, median imputation assumptions, lack of external clinical validation, and linear decision boundaries.
Future Enhancements: Implement advanced resampling techniques (e.g., SMOTE), optimal probability threshold tuning, regression-based BMI imputation, feature interactions (such as `age × hypertension`), tree ensemble comparisons (XGBoost/Random Forest), and explainability tooling (SHAP values).
